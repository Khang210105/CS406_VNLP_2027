```python
from pathlib import Path
from PIL import Image

input_dir = Path("../dataset/raw/view1/images")
output_dir = Path("../dataset/raw/view1/images")
output_dir.mkdir(parents=True, exist_ok=True)

count = 0
for path in input_dir.iterdir():
    if path.suffix.lower() not in {".jpg", ".jpeg", ".png", ".bmp", ".webp"}:
        continue
    with Image.open(path) as img:
        if img.size != (1920, 1080):
            print(f"Bỏ qua {path.name}: kích thước {img.size}")
            continue
        img.crop((0, 0, 1310, 1080)).save(output_dir / path.name)
        count += 1

print(f"Đã cắt {count} ảnh, kích thước mới: 1310 × 1080")
```
