```python
from pathlib import Path
import cv2
import numpy as np
from PIL import Image

root = Path("../dataset/raw")

for i in range(1, 5):
    folder = root / f"view{i}" / "images"
    print(f"\n--- view{i} ---")

    if not folder.exists():
        print("Không tìm thấy thư mục:", folder)
        continue

    files = [p for p in folder.rglob("*") if p.is_file()]
    print("Tổng số file:", len(files))

    for path in files[:3]:
        try:
            data = np.fromfile(str(path), dtype=np.uint8)
            img_cv = cv2.imdecode(data, cv2.IMREAD_COLOR)

            with Image.open(path) as img_pil:
                print(
                    f"{path.name}\n"
                    f"  Pillow: {img_pil.format}, {img_pil.size}\n"
                    f"  OpenCV đọc được: {img_cv is not None}"
                )
        except Exception as e:
            print(path.name, "-> Lỗi:", repr(e))
```
